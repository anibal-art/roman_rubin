```python
from pathlib import Path
import json

fname = Path("metrics_BH-FFP.ipynb")
outname = fname.with_name(fname.stem + "_cleaned.ipynb")

with open(fname, "r", encoding="utf-8", errors="ignore") as f:
    lines = f.readlines()

cleaned = []
mode = "normal"   # normal | keep_head | skip_other

for line in lines:
    if line.startswith("<<<<<<< "):
        mode = "keep_head"
        continue
    elif line.startswith("=======") and mode == "keep_head":
        mode = "skip_other"
        continue
    elif line.startswith(">>>>>>> ") and mode in ("keep_head", "skip_other"):
        mode = "normal"
        continue

    if mode in ("normal", "keep_head"):
        cleaned.append(line)

with open(outname, "w", encoding="utf-8") as f:
    f.writelines(cleaned)

print(f"Archivo generado: {outname}")

# Verificación JSON
try:
    with open(outname, "r", encoding="utf-8") as f:
        json.load(f)
    print("El notebook resultante es JSON válido.")
except Exception as e:
    print("El archivo limpio todavía no es JSON válido:")
    print(repr(e))
```
