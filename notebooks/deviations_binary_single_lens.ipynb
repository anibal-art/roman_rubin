# Fits

Relevant Function: 

fit_rubin_roman(Source,event_params, path_to_save_fit, path_ephemerides,model,algo,origin,
                               band_roman, band_rubin_u, band_rubin_g, band_rubin_r,
                                           band_rubin_i, band_rubin_z, band_rubin_y)

This function performs the fitting of microlensing events using data from both the Rubin Observatory (LSST) and the Roman Space Telescope (WFIRST). It supports various microlensing models, including Finite-Source Point Lens (FSPL), Uniform Source Binary Lens (USBL), and Point Source Point Lens (PSPL). Additionally, the function accommodates multiple fitting algorithms, such as Trust Region Reflective (TRF), Markov Chain Monte Carlo (MCMC), and Differential Evolution (DE).

The boundaries used for the fitting process can be modified within the function's script.

```python
# i=0
for i in range(1,100):
    info_dataset, pyLIMA_parameters, bands = read_data(path_to_save_model+'/Event_'+str(int(i))+'.h5')
    algo='TRF'
    
    Source = i
    rango = 1
    
    lc_to_fit = {}
    for telo in bands:
        if not len(bands[telo]['mag'])==0:
            df = bands[telo][['time', 'mag','err_mag']].to_pandas()
            lc_to_fit[telo] = df.values
        else:
            lc_to_fit[telo] = []
    origin = info_dataset[2]
    
    fit_usbl, event_fit_usbl, pyLIMAmodel_usbl = fit_rubin_roman(f"USBL_{int(i)}_",pyLIMA_parameters, path_to_save_fit, path_ephemerides,"USBL_NoPiE",algo,origin, rango,
                               lc_to_fit["W149"], lc_to_fit["u"], lc_to_fit["g"], lc_to_fit["r"],
                                           lc_to_fit["i"], lc_to_fit["z"],lc_to_fit["y"])
    
    fit_pspl, event_fit_pspl, pyLIMAmodel_pspl = fit_rubin_roman(f"PSPL_{int(i)}_",pyLIMA_parameters, path_to_save_fit, path_ephemerides,'PSPL_NoPiE',algo,origin, rango,
                               lc_to_fit["W149"], lc_to_fit["u"], lc_to_fit["g"], lc_to_fit["r"],
                                           lc_to_fit["i"], lc_to_fit["z"],lc_to_fit["y"])
```

```python
import numpy as np
import statsmodels.api as sm
from scipy import stats
ll1 =-fit_usbl.likelihood_photometry(fit_usbl.fit_results['best_model'])
ll0 =-fit_pspl.likelihood_photometry(fit_pspl.fit_results['best_model'])

# Likelihood Ratio Test
LR_statistic = 2 * (ll1 - ll0)
df = len(fit_usbl.fit_results['best_model']) - len(fit_pspl.fit_results['best_model'])  # Difference in parameters
p_value = 1 - stats.chi2.cdf(LR_statistic, df)

print(f"LR Statistic: {LR_statistic}")
print(f"Degrees of Freedom: {df}")
print(f"P-value: {p_value}")

if p_value < 0.05:
    print("Reject Null Model (Complex model is better)")
else:
    print("Fail to Reject Null Model")
```

```python
k_usbl = len(fit_usbl.fit_results['best_model'])
n_usbl = sum([len(telo.lightcurve) for telo in event_fit_usbl.telescopes])
penalty_term = k_usbl*np.log(n_usbl)
BIC_usbl = ll1 + penalty_term
print(BIC_usbl)
```

```python

k_pspl = len(fit_pspl.fit_results['best_model'])
n_pspl = sum([len(telo.lightcurve) for telo in event_fit_pspl.telescopes])
penalty_term_pspl = k_pspl*np.log(n_pspl)
BIC_pspl = ll1 + penalty_term_pspl
print(BIC_pspl)
```

```python
print(BIC_usbl)
print(BIC_pspl)
if BIC_usbl>BIC_pspl:
    print('Model PSPL is better')
else:
    print('Model USBL is better')
```

```python
#we can make some tests to check if this is a good event before make the plot
from scipy.stats import chi2
chi2_roman = fit_roman.fit_results['chi2']
chi2_rr = fit_rr.fit_results['chi2']
DOF_roman = len(pyLIMAmodel_roman.event.telescopes[0].lightcurve['time'])-len(fit_roman.fit_results['best_model'])
DOF_rr = sum([len(pyLIMAmodel_rr.event.telescopes[i].lightcurve['time']) for i in range(len(pyLIMAmodel_rr.event.telescopes))])-len(fit_rr.fit_results['best_model'])
print('Chi square reduced for Roman', chi2_roman/DOF_roman)
print('Chi square reduced for RR', chi2_rr/DOF_rr)
print('p-value of a chi squared test', chi2.sf(chi2_roman, DOF_roman))
# #metrics

#metric alpha
leng = len(fit_roman.fit_results['best_model'])-2
true_array = np.array(true_params)[0:leng]
fit_rr_array = fit_rr.fit_results['best_model'][0:leng]
fit_roman_array = fit_roman.fit_results['best_model'][0:leng]
sigma_rr_array = np.sqrt(np.diag(fit_rr.fit_results['covariance_matrix']))[0:leng]
sigma_roman_array = np.sqrt(np.diag(fit_roman.fit_results['covariance_matrix']))[0:leng]

alpha_roman = abs(true_array-fit_roman_array)/true_array
alpha_rr = abs(true_array-fit_rr_array)/true_array
#metric beta
beta_roman = abs(true_array-fit_roman_array)/sigma_roman_array
beta_rr = abs(true_array-fit_rr_array)/sigma_rr_array
#metric gamma
gamma_roman = sigma_roman_array/fit_roman_array
gamma_rr = sigma_rr_array/fit_rr_array
```
