```python
import numpy as np
import matplotlib.pyplot as plt


path = '/home/anibal-pc/microlensing/simulation_Rubin/roman_rubin/ephemerides/'
path_ephemerides = path+'Roman_positions.npy'
eph = np.load(path+'Roman_positions.npy')
eph2 = np.loadtxt(path+'Gaia.txt')
```

```python
# Extraer columnas
time = eph[:, 0]
x = eph[:, 1]
y = eph[:, 2]
z = eph[:, 3]  # Suponiendo que esta es una variable relevante

# Gráfico 2D: X vs. Y
plt.figure(figsize=(8, 6))
plt.plot(x, y, marker='o', linestyle='-', color='b')
plt.xlabel('X Position')
plt.ylabel('Y Position')
plt.title('Satellite Ephemerides: X vs. Y')
plt.grid()
plt.show()

# Gráfico 3D: X, Y, Z
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')
ax.plot(x, y, z, marker='o', linestyle='-', color='r')
ax.set_xlabel('X Position')
ax.set_ylabel('Y Position')
ax.set_zlabel('Z (or additional parameter)')
ax.set_title('Satellite Ephemerides in 3D')
plt.show()

# Gráfico de evolución temporal
plt.figure(figsize=(8, 6))
plt.plot(time, z, marker='o', linestyle='-', color='g')
plt.xlabel('Time')
plt.ylabel('Z (or additional parameter)')
plt.title('Evolution of Z over Time')
plt.grid()
plt.show()
```
